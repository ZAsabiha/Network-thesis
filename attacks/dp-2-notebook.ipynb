```python
import pandas as pd
import os

# The path from kagglehub
path = "/root/.cache/kagglehub/datasets/badcodebuilder/insdn-dataset/versions/1"  # or whatever path was printed

# If you didn't save the path variable, run this:
import kagglehub
path = kagglehub.dataset_download("badcodebuilder/insdn-dataset")
print(f"Dataset path: {path}")

# List all files in the dataset
print(f"\nFiles in dataset:")
files = os.listdir(path)
for file in files:
    print(f"  - {file}")
    file_path = os.path.join(path, file)
    if os.path.isfile(file_path) and file.endswith('.csv'):
        # Check file size
        size = os.path.getsize(file_path) / (1024 * 1024)  # MB
        print(f"    Size: {size:.2f} MB")

# Load the CSV files
csv_files = [f for f in files if f.endswith('.csv')]
df_dict = {}

for file in csv_files:
    file_path = os.path.join(path, file)
    df = pd.read_csv(file_path)
    df_dict[file] = df
    print(f"\n{file}:")
    print(f"  - Rows: {len(df)}")
    print(f"  - Columns: {len(df.columns)}")
    print(f"  - Column names (first 10): {df.columns[:10].tolist()}")
    print(f"  - Data types:")
    print(f"    {df.dtypes.value_counts()}")
    print(f"  - First 2 rows:")
    print(df.head(2))
    print(f"  - Sample values from first row: {df.iloc[0].tolist()[:5]}")

# Combine all dataframes
if df_dict:
    all_dfs = []
    for name, df in df_dict.items():
        df_copy = df.copy()
        df_copy['source'] = name.replace('.csv', '')
        all_dfs.append(df_copy)
    
    combined_df = pd.concat(all_dfs, ignore_index=True)
    print(f"\n{'='*50}")
    print(f"Combined dataset:")
    print(f"  - Total rows: {len(combined_df)}")
    print(f"  - Total columns: {len(combined_df.columns)}")
    
    # Check for label/target column
    label_cols = [col for col in combined_df.columns if 'label' in col.lower() or 'class' in col.lower()]
    if label_cols:
        print(f"\n  - Potential label columns: {label_cols}")
        for col in label_cols:
            print(f"    {col} distribution:")
            print(f"    {combined_df[col].value_counts()}")
    
    # Save for later use
    df = combined_df
    print(f"\n✅ Dataset loaded and ready for training!")
    print(f"📊 Shape: {df.shape}")
    print(f"📋 Columns: {df.columns.tolist()}")
```

```python
import pandas as pd

p_normal = "/kaggle/input/datasets/badcodebuilder/insdn-dataset/InSDN_DatasetCSV/Normal_data.csv"
p_meta   = "/kaggle/input/datasets/badcodebuilder/insdn-dataset/InSDN_DatasetCSV/metasploitable-2.csv"
p_ovs    = "/kaggle/input/datasets/badcodebuilder/insdn-dataset/InSDN_DatasetCSV/OVS.csv"

for name, p in [("Normal", p_normal), ("Metasploitable", p_meta), ("OVS", p_ovs)]:
    df = pd.read_csv(p, nrows=3)
    print(f"\n=== {name} ({df.shape[1]} cols) ===")
    print(list(df.columns))
```

```python
df_normal = pd.read_csv(p_normal, low_memory=False)
df_meta   = pd.read_csv(p_meta, low_memory=False)
df_ovs    = pd.read_csv(p_ovs, low_memory=False)

for d in (df_normal, df_meta, df_ovs):
    d.columns = [c.strip() for c in d.columns]

df_all = pd.concat([df_normal, df_meta, df_ovs], ignore_index=True)
print(df_all.shape)
print(df_all['Label'].value_counts())
```

```python
def map_label_to_class(raw_label):
    raw = str(raw_label).strip().lower()
    if raw in ("normal", "benign"):
        return "Normal"
    if "ddos" in raw:
        return "DDoS"
    if "dos" in raw and "ddos" not in raw:
        return "DoS"
    if "probe" in raw or "scan" in raw:
        return "Probe"
    if "bfa" in raw or "brute" in raw:
        return "BFA"
    if "web" in raw:
        return "Web-Attack"
    if "bot" in raw:
        return "Botnet"
    if "u2r" in raw or "exploit" in raw:
        return "U2R"
    return "Unknown"

df_all['class'] = df_all['Label'].apply(map_label_to_class)
print(df_all['class'].value_counts())
print("Unknown labels found:", df_all[df_all['class'] == 'Unknown']['Label'].unique())
```

```python
#feature engineering

import numpy as np

work = df_all.copy()
work['duration_sec'] = (work['Flow Duration'] / 1_000_000.0).clip(lower=0.001)
work['packet_count'] = work['Tot Fwd Pkts'] + work['Tot Bwd Pkts']
work['byte_count']   = work['TotLen Fwd Pkts'] + work['TotLen Bwd Pkts']
work['protocol']     = work['Protocol']
work['avg_packet_size'] = work['byte_count'] / work['packet_count'].replace(0, np.nan)
work['packet_rate']     = work['packet_count'] / work['duration_sec']
work['byte_rate']       = work['byte_count'] / work['duration_sec']

feature_cols = ['duration_sec', 'packet_count', 'byte_count',
                 'avg_packet_size', 'packet_rate', 'byte_rate', 'protocol']

work = work.replace([np.inf, -np.inf], np.nan)
before = len(work)
work = work.dropna(subset=feature_cols + ['class'])
work = work[work['class'] != 'Unknown']
print(f"Dropped {before - len(work)} rows. Remaining: {len(work)}")

X = work[feature_cols]
y = work['class']
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from imblearn.over_sampling import SMOTE
import joblib

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

# k_neighbors capped low because U2R has very few samples in the training split
sm = SMOTE(random_state=42, k_neighbors=3)
X_train_res, y_train_res = sm.fit_resample(X_train, y_train)
print("After SMOTE:", y_train_res.value_counts())

clf = RandomForestClassifier(
    n_estimators=200, class_weight="balanced", random_state=42, n_jobs=-1
)
clf.fit(X_train_res, y_train_res)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, f1_score

y_pred = clf.predict(X_test)

print("=== Classification Report ===")
print(classification_report(y_test, y_pred, zero_division=0))

macro_f1 = f1_score(y_test, y_pred, average='macro')
weighted_f1 = f1_score(y_test, y_pred, average='weighted')
print(f"\nMacro-F1 (treats all classes equally): {macro_f1:.4f}")
print(f"Weighted-F1 (accounts for class size):  {weighted_f1:.4f}")

print("\n=== Confusion Matrix ===")
labels_sorted = sorted(y.unique())
print("Classes:", labels_sorted)
print(confusion_matrix(y_test, y_pred, labels=labels_sorted))
```

```python
from imblearn.over_sampling import SMOTE

# Only meaningfully oversample classes with enough real diversity to interpolate safely.
# Leave Web-Attack and U2R alone - too few real points, SMOTE distorts more than it helps.
# class_weight='balanced' in the RF still gives them a boost during training.
sampling_strategy = {
    "BFA": 8000,      # up from ~1050 real training samples
    "Botnet": 3000,   # up from ~120
}

sm2 = SMOTE(random_state=42, k_neighbors=3, sampling_strategy=sampling_strategy)
X_train_res2, y_train_res2 = sm2.fit_resample(X_train, y_train)
print(y_train_res2.value_counts())

clf2 = RandomForestClassifier(
    n_estimators=200, class_weight="balanced", random_state=42, n_jobs=-1
)
clf2.fit(X_train_res2, y_train_res2)

y_pred2 = clf2.predict(X_test)
from sklearn.metrics import classification_report, f1_score
print(classification_report(y_test, y_pred2, zero_division=0))
print("Macro-F1:", f1_score(y_test, y_pred2, average='macro'))
```
